```python
import pandas as pd
from local.constants import WORKSPACE_ROOT
```

```python
df = pd.read_csv(WORKSPACE_ROOT/"data/lipids/27Aug2025/Epi300_size.txt", sep="\t", encoding="latin1")
print(df.shape)
df.head(2)
```

```python
# EPI
SCADC_SAMPLES = [
    'SCADC-1 1', 'SCADC-1 2', 'SCADC-1 3',
    'SCADC1 1', 'SCADC1 2', 'SCADC1 3',
    'Epi300 1', 'Epi300 2', 'Epi300 3'
]
df[df["Sample Name"].isin(SCADC_SAMPLES)]
```

```python
dfs = pd.read_csv(WORKSPACE_ROOT/"data/lipids/27Aug2025/Size_Total_27Aug2025", sep="\t", encoding="latin1")
dfs
```

```python
from dataclasses import dataclass
@dataclass
class Entry:
    name: str
    rep: str
    size: float
    fraction: float

data: list[Entry] = []
names = set()

for _, r in dfs.iterrows():
    n = r["Sample Name"]
    toks = n.split(" ")
    rep = toks[-1]
    name = " ".join(toks[:-1])
    name = {
        "SCADC-1": "Clone2",
        "SCADC_D4_clone1": "Clone1",
        "Epi300_pCC1": "PCC1",
        "Epi300": "EPI300",
    }.get(name, name)
    
    raw_vals = r[[c for c in dfs.columns if "Pk" in c]]
    sizes = raw_vals[:3]
    frac = raw_vals[3:]
    vals = []
    for s, f in zip(sizes, frac):
        if s > 1000: continue # dust?
        if f == 0: continue
        vals.append((s, f))
    total = sum(f for s, f in vals)
    vals = [Entry(name, rep, s, f/total) for s, f in vals]
    data+=vals
    names.add(name)
```

```python
_order = [
    "Clone1",
    "Clone2",
    "PCC1",
    "EPI300",
]

names = sorted(list(names), key=lambda n: _order.index(n))
data = sorted(data, key=lambda e: (_order.index(e.name), e.rep))
names, data
```

```python
from local.figures.template import BaseFigure, ApplyTemplate, go
from local.figures.colors import Palettes, COLORS, Color
import numpy as np

def _to_name(e: Entry):
    return f"{e.name} {e.rep}"

def _scale_size(s):
    return np.sqrt(s)*40

palette = Palettes.PLOTLY[:2]+[Color.Hex("555555"), Color.Hex("999999")]
colors = {}
fig = BaseFigure(
    shape=(2, 1),
    column_widths=[5, 1],
    shared_yaxes=False,
)
for i, name in enumerate(names):
    color = palette[i]
    colors[name] = color
    fig.add_trace(
        go.Scatter(
            x = [_to_name(e) for e in data if e.name == name],
            y = [e.size for e in data if e.name == name],
            mode="markers",
            marker=dict(
                color=color.color_value, size=[_scale_size(e.fraction) for e in data if e.name == name],
                line=dict(color=COLORS.BLACK),
            ),
            showlegend=False,
        ),
        row=1, col=1
    )

size_ladder = [0.33, 0.67, 1.0]
fig.add_trace(
    go.Scatter(
        x = [0]*len(size_ladder),
        y = [i for i in range(len(size_ladder))],
        text = [f"{s:.0%}" for s in size_ladder],
        mode="markers+text",
        textposition="middle right",
        marker=dict(
            size=[_scale_size(s) for s in size_ladder],
            color=COLORS.TRANSPARENT,
            line=dict(color=COLORS.BLACK),
        ),
        showlegend=False,
    ),
    row=1, col=2,
)
invis = dict(showticklabels=False, linecolor=COLORS.TRANSPARENT, ticks=None)
fig = ApplyTemplate(
    fig,
    axis = {
        "1 1 y": dict(title="mean particle size (nm)", dtick=30),
        # "1 1 y": dict(title="mean particle size (nm)", type="log", range=[1, np.log10(500)]),
        "1 1 x": dict(
            ticks=None, tickvals=[_to_name(e) for e in data],
            ticktext=[f'<span style="color:{colors[e.name].AsHex()}">{e.name}</span>' if e.rep=="2" else "" for i, e in enumerate(data)],
        ),
        "2 1 x": invis|dict(range=[-1, 3]),
        "2 1 y": invis|dict(range=[-5, len(size_ladder)]),
    },
    layout=dict(
        width=800, height=500,
    )
)
fig.write_image("./cache/DLS_peaks.png")
fig.show()
```
